```python
b = {}

for w in name:
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1 , ch2 in zip(chs, chs[1:]):
        bigram = (ch1, ch2)
        b[bigram] = b.get(bigram, 0) + 1
        # print(ch1, ch2)
        
```

```python
N = torch.zeros((28, 28), dtype=torch.int32)
chars = sorted(list(set(''.join(name))))
stoi = {s:i for i,s in enumerate(chars)}
stoi['S'] = 26
stoi['E'] = 27
```

```python

for w in name:
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1 , ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1
        
```
