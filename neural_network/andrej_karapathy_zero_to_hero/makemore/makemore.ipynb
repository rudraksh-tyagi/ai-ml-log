```python
name = pd.read_csv('/home/entr0py/Downloads/Indian_Names.csv')
name[:10]
len(name)
min(len(w) for w in name)
max(len(w) for w in name)
```

```python
print(name.columns.tolist())
print(name.isnull().sum())
name = name.dropna(subset=["Name"])

names = name["Name"].str.lower().tolist()
```

```python
print(len(names))
print(names[:20])
```

```python
b = {}
for w in names:
    chs = ['<S>'] + list(w) + ['<E>']   
    for ch1, ch2 in zip(chs, chs[1:]):
        bigram = (ch1, ch2)
        b[bigram] = b.get(bigram, 0) + 1
        print(ch1, ch2)
```

```python
sorted(b.items() , key = lambda kv : -kv[1])
```
