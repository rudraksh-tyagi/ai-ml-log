# Make More — Indian Names Bigram Model

This notebook cleans the names dataset, builds a character vocabulary, counts adjacent-character pairs (bigrams), and stores those counts in a PyTorch tensor.

**Run the notebook from top to bottom.** The cleaning step is deliberately placed immediately after loading the CSV so later cells always use the cleaned names.


## 1. Imports

```python
import string
from pathlib import Path

import pandas as pd
import torch
```

## 2. Load the dataset

The notebook first checks the original path used on this machine, then checks for `Indian_Names.csv` in the notebook's working directory.

```python
csv_path = Path("/home/entr0py/Downloads/Indian_Names.csv")

if not csv_path.exists():
    csv_path = Path("Indian_Names.csv")

if not csv_path.exists():
    raise FileNotFoundError(
        "Could not find Indian_Names.csv. Put it in the notebook's working "
        "directory or update csv_path to the correct location."
    )

data = pd.read_csv(csv_path)

if "Name" not in data.columns:
    raise KeyError(f"Expected a 'Name' column. Found: {data.columns.tolist()}")

print("Dataset shape:", data.shape)
print("Columns:", data.columns.tolist())
data.head()
```

## 3. Clean the names

Keep only English letters `a-z`, convert to lowercase, and discard empty names. This cell creates `names` from the loaded CSV, so it cannot accidentally clean an old or undefined variable.

```python
raw_names = data["Name"].dropna().astype(str).str.lower().tolist()

names = [
    "".join(ch for ch in word if ch in string.ascii_lowercase)
    for word in raw_names
]
names = [word for word in names if word]

print("Names retained:", len(names))
print("Sample cleaned names:", names[:20])
```

```python
b = {}
for w in names:
    chs = ['<S>'] + list(w) + ['<E>']   
    for ch1, ch2 in zip(chs, chs[1:]):
        bigram = (ch1, ch2)
        b[bigram] = b.get(bigram, 0) + 1
        print(ch1, ch2)
```
