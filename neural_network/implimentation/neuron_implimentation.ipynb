```python
import numpy as np

X = np.array([[2.0, 3.0]])
y = np.array([[1.0]])

W = np.array([
    [0.5, 0.2],
    [-0.3, 0.8]
])

b = np.array([[1.0, 0.5]])

W2 = np.array([
    [0.4],
    [-0.2]
])

b2 = np.array([[0.5]])

lr = 0.1

for epoch in range(1000):

    # ---------- FORWARD ----------
    Z = X @ W + b
    A = np.maximum(0, Z)

    Z2 = A @ W2 + b2

    # ---------- LOSS ----------
    loss = np.mean((Z2 - y) ** 2)

    # ---------- BACKPROP ----------
    dZ2 = 2 * (Z2 - y)

    dW2 = A.T @ dZ2
    db2 = dZ2

    dA = dZ2 @ W2.T

    dZ = dA * (Z > 0)

    dW = X.T @ dZ
    db = dZ

    # ---------- UPDATE ----------
    W = W - lr * dW
    b = b - lr * db

    W2 = W2 - lr * dW2
    b2 = b2 - lr * db2

    if epoch % 100 == 0:
        print("Epoch:", epoch, "Loss:", loss)

print("Final prediction:", Z2)
```

```python
import numpy as np

# ---------- DATA ----------
X = np.array([[2.0, 3.0]])
y = np.array([[1.0]])

# ---------- PARAMETERS ----------
W1 = np.array([
    [0.5, 0.2],
    [-0.3, 0.8]
])

b1 = np.array([[1.0, 0.5]])

W2 = np.array([
    [0.4],
    [-0.2]
])

b2 = np.array([[0.5]])

lr = 0.1

# ---------- TRAINING ----------
for epoch in range(1000):

    # ===== FORWARD =====
    Z1 = X @ W1 + b1
    A1 = np.maximum(0, Z1)

    Z2 = A1 @ W2 + b2

    # ===== LOSS =====
    loss = np.mean((Z2 - y) ** 2)

    # ===== BACKWARD =====

    # Output layer
    dZ2 = 2 * (Z2 - y)

    dW2 = A1.T @ dZ2
    db2 = dZ2

    # Send gradient to previous layer
    dA1 = dZ2 @ W2.T

    # ReLU
    dZ1 = dA1 * (Z1 > 0)


    dW2 = A1.T @ dZ2
    db2 = dZ2

    # Send gradient to previous layer
    dA1 = dZ2 @ W2.T

    # ReLU
    # First layer
    dW1 = X.T @ dZ1
    db1 = dZ1

    # ===== UPDATE =====
    W2 -= lr * dW2
    b2 -= lr * db2

    W1 -= lr * dW1
    b1 -= lr * db1

    if epoch % 100 == 0:
        print(epoch, loss)

print("Prediction:", Z2)
```
