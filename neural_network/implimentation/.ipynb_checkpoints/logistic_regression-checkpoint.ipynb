```python
import numpy as np

x = np.array([[-2], [-0.5], [0], [0.5], [2]])
y= np.array([[0] , [0] , [0] ,[1],[1],[1]])
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
# w = np.random.randn(1, 1)
# b = np.zeros((1, 1))
z = x @ w + b
probabilities = sigmoid(z)
y_pred = sigmoid(z)
print("Predicted probabilities:", probabilities)
```

```python
loss = -np.mean(
    y * np.log(y_pred) +
    (1 - y) * np.log(1 - y_pred)
)
```

```python
N = len(x)

error  =   y_pred - y

dw = (1/N) * (x.T @ error)
db = (1/N) * np.sum(error)

lr = 0.01
```

```python

for epoch in range(1000):

    # 1. Predict
    z = x @ w + b
    y_pred = sigmoid(z)

    # 2. Calculate loss
    loss = -np.mean(
        y * np.log(y_pred) +
        (1-y) * np.log(1-y_pred)
    )

    # 3. Calculate gradients
    error = y_pred - y

    dw = (1/N) * (x.T @ error)
    db = (1/N) * np.sum(error)

    # 4. Update parameters
    w = w - lr * dw
    b = b - lr * db

    if epoch % 100 == 0:
        print(epoch, loss)

print("w =", w)
print("b =", b)
```
