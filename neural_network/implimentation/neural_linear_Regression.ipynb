```python
import numpy as np
```

```python
np.random.seed(42)
```

```python
x = np.random.randn(100,1)
y = 2*x +1 + 0.1*np.random.randn(100,1)
y
```

```python
w = np.random.randn(1,1)
b= np.zeros((1,1))
```

```python
y_pred = x@w+b
print(y_pred.shape)
print(y_pred[:5])
```

```python
loss = np.mean((y_pred-y)**2)
print(loss)
```

```python
N = len(x)
error = y_pred - y

dw = (2/N)*x.T@error
db = (2/N)*np.sum(error)
```

```python
lr = 0.01
```

```python
w = w - lr*dw
b = b - lr*db
```

```python
y_pred = x @ w + b
loss = np.mean((y_pred - y) ** 2)

print(loss)
```

```python
for epoch in range(1000):
    y_pred = x @ w + b

    error = y_pred - y
    dw = (2/N)*x.T@error
    db = (2/N)*np.sum(error)
    w = w - lr * dw
    b = b - lr * db
if epoch % 100 == 0:
    loss = np.mean(error ** 2)
    print(epoch, loss)

print(w)
print(b)
```
